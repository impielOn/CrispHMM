```python
import numpy as np
import os

# ------------------------------------------------------------
# 1. Load gRNA data from the Excel sheet 'hela'
# ------------------------------------------------------------
file_path = '13059_2018_1459_MOESM5_ESM.xlsx'
df_grna = pd.read_excel(file_path, sheet_name='hela')
print(f"Loaded {len(df_grna)} gRNAs")
print(df_grna.head())

# Keep only the columns we need
df_grna = df_grna[['Chromosome', 'Start', 'Normalized efficacy']].copy()

# ------------------------------------------------------------
# 2. Parse binary output files
# ------------------------------------------------------------
BIN_SIZE = 10000  # must match the -b parameter used in BinarizeBed

# The order of marks (from the second line of any binary file)
# We extracted from HeLa_chr1_binary.txt: H3K27ac H3K27me3 H3K36me3 H3K4me1 H3K4me3
mark_order = ['H3K27ac', 'H3K27me3', 'H3K36me3', 'H3K4me1', 'H3K4me3']

# For each chromosome, store a list of binary strings (each string length = number of marks)
chrom_binaries = {}

for filename in os.listdir('binary_output'):
    if not filename.startswith('HeLa_chr') or not filename.endswith('_binary.txt'):
        continue
    # Extract chromosome name: e.g., 'HeLa_chr1_binary.txt' -> 'chr1'
    chrom = filename.replace('HeLa_', '').replace('_binary.txt', '')
    filepath = os.path.join('binary_output', filename)

    with open(filepath, 'r') as f:
        lines = f.readlines()
    # Skip first line (cell and chrom), second line is mark names (we use our fixed order)
    # Lines 3.. end: each line is a tab-separated list of 0/1
    binaries = []
    for line in lines[2:]:
        line = line.strip()
        if not line:
            continue
        # Join all values into one string (e.g., "01001")
        bin_str = ''.join(line.split('\t'))
        if len(bin_str) != len(mark_order):
            # Safety: if length mismatch, warn but continue
            print(f"Warning: {chrom} line length {len(bin_str)} != {len(mark_order)}")
        binaries.append(bin_str)
    chrom_binaries[chrom] = binaries
    print(f"Loaded {chrom}: {len(binaries)} bins")

# ------------------------------------------------------------
# 3. Map each gRNA to its bin and extract binary features
# ------------------------------------------------------------
def get_bin_features(chrom, start_1based):
    """Return binary string (0/1) for the 10kb bin containing start_1based."""
    if chrom not in chrom_binaries:
        return None
    pos0 = start_1based - 1          # convert 1‑based to 0‑based
    bin_idx = pos0 // BIN_SIZE
    if bin_idx < len(chrom_binaries[chrom]):
        return chrom_binaries[chrom][bin_idx]
    else:
        return None

df_grna['bin_features'] = df_grna.apply(
    lambda r: get_bin_features(r['Chromosome'], r['Start']), axis=1
)

# Drop gRNAs without a bin (should be none if chromosomes match)
before = len(df_grna)
df_grna = df_grna.dropna(subset=['bin_features'])
print(f"Dropped {before - len(df_grna)} gRNAs with no matching bin")

# ------------------------------------------------------------
# 4. Expand the binary string into separate mark columns
# ------------------------------------------------------------
for i, mark in enumerate(mark_order):
    df_grna[mark] = df_grna['bin_features'].apply(lambda s: int(s[i]) if s is not None else None)

# ------------------------------------------------------------
# 5. Save the final feature table
# ------------------------------------------------------------
output_cols = ['Chromosome', 'Start', 'Normalized efficacy'] + mark_order
df_grna[output_cols].to_csv('gRNA_10kb_binary_features.csv', index=False)
print("Saved to gRNA_10kb_binary_features.csv")
print(df_grna[output_cols].head())
```
